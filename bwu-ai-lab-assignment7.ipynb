**WAP a python program to implement the minmax algo for a two-player game .**

```python
# Minimax Algorithm with User Input

def minimax(depth, index, is_max, values):
    if depth == 3:   # 8 leaf nodes → height = 3
        return values[index]

    if is_max:
        return max(
            minimax(depth + 1, index * 2, False, values),
            minimax(depth + 1, index * 2 + 1, False, values)
        )
    else:
        return min(
            minimax(depth + 1, index * 2, True, values),
            minimax(depth + 1, index * 2 + 1, True, values)
        )

# Take input from user
values = list(map(int, input("Enter 8 leaf node values (space-separated): ").split()))

# Validate input
if len(values) != 8:
    print("Please enter exactly 8 values.")
else:
    optimal_score = minimax(0, 0, True, values)
    print("Optimal value for MAX player is:", optimal_score)
```

**Write a python program to implement the Alpha-Beta pruning algorithm for a two player game.Use the leaf-node values 3,5,2,9,12,5,23,23. Determine the optimal score for the max player and display which branches are pruned during the search**

```python
class AlphaBetaPruning:
    def __init__(self, values):
        self.leaf_values = values
        self.leaf_index = 0
        self.pruned_branches = []

    def minimax(self, depth, node_index, maximizing_player, alpha, beta):
        # Base case: Leaf node reached
        if depth == 3:
            value = self.leaf_values[self.leaf_index]
            self.leaf_index += 1
            return value

        if maximizing_player:
            max_eval = float('-inf')
            # Assuming a binary tree (2 children per node)
            for i in range(2):
                eval_val = self.minimax(depth + 1, node_index * 2 + i, False, alpha, beta)
                max_eval = max(max_eval, eval_val)
                alpha = max(alpha, eval_val)
                
                # Beta cutoff
                if beta <= alpha:
                    # Record the remaining sibling/branches that will be pruned
                    # In a standard binary tree, if we prune on i == 0, the next branch (i == 1) is pruned.
                    pruned_node_info = f"Depth {depth+1}, Node {node_index * 2 + (1 - i)}"
                    self.pruned_branches.append(pruned_node_info)
                    break
            return max_eval
        
        else:
            min_eval = float('inf')
            for i in range(2):
                eval_val = self.minimax(depth + 1, node_index * 2 + i, True, alpha, beta)
                min_eval = min(min_eval, eval_val)
                beta = min(beta, eval_val)
                
                # Alpha cutoff
                if beta <= alpha:
                    pruned_node_info = f"Depth {depth+1}, Node {node_index * 2 + (1 - i)}"
                    self.pruned_branches.append(pruned_node_info)
                    break
            return min_eval

# Leaf node values provided
values = [3, 5, 2, 9, 12, 5, 23, 23]

# Initialize and run the algorithm
ab_game = AlphaBetaPruning(values)
optimal_score = ab_game.minimax(depth=0, node_index=0, maximizing_player=True, alpha=float('-inf'), beta=float('inf'))

# Display Results
print("=== Alpha-Beta Pruning Results ===")
print(f"Leaf-node values: {values}")
print(f"Optimal Score for the MAX player: **{optimal_score}**\n")

print("Branches Pruned During Search:")
if ab_game.pruned_branches:
    for branch in ab_game.pruned_branches:
        print(f" - Pruned at {branch}")
else:
    print(" - No branches were pruned.")
```
